````python
dico={}
dico['protein']='proteine'
dico['DNA']='ADN'
dico['RNA']='ARN'
print(dico)
````

````python
res1 = dico['DNA']
print(res1)
#res2 = dico['ADN'] # KeyError: 'ADN'
res2 = dico['ADN']
````

Attention, on peut seulement obtenir la valeur à partir de la clé et pas l'inverse.
